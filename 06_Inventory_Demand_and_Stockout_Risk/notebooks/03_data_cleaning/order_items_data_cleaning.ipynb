```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/order_items.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["fulfillment_status"] = pd.to_datetime(df["fulfillment_status"], errors="coerce")  # Convert date values safely.
df["quantity"] = pd.to_numeric(df["quantity"], errors="coerce")  # Standardise numeric values.
df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")  # Standardise numeric values.
df["discount_amount"] = pd.to_numeric(df["discount_amount"], errors="coerce")  # Standardise numeric values.
df["unit_cost"] = pd.to_numeric(df["unit_cost"], errors="coerce")  # Standardise numeric values.
df["order_item_id"] = df["order_item_id"].astype("string").str.strip()  # Standardise text values.
df["order_id"] = df["order_id"].astype("string").str.strip()  # Standardise text values.
df["product_id"] = df["product_id"].astype("string").str.strip()  # Standardise text values.
df["fulfillment_status"] = df["fulfillment_status"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["order_item_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/order_items_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/order_items_quality_profile.csv", index=False)  # Save quality profile.
```
