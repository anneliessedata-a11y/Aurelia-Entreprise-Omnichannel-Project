```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/stores.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["store_format"] = pd.to_datetime(df["store_format"], errors="coerce")  # Convert date values safely.
df["opening_date"] = pd.to_datetime(df["opening_date"], errors="coerce")  # Convert date values safely.
df["closing_date"] = pd.to_datetime(df["closing_date"], errors="coerce")  # Convert date values safely.
df["store_status"] = pd.to_datetime(df["store_status"], errors="coerce")  # Convert date values safely.
df["closing_date"] = pd.to_numeric(df["closing_date"], errors="coerce")  # Standardise numeric values.
df["floor_area_sqm"] = pd.to_numeric(df["floor_area_sqm"], errors="coerce")  # Standardise numeric values.
df["store_id"] = df["store_id"].astype("string").str.strip()  # Standardise text values.
df["store_name"] = df["store_name"].astype("string").str.strip()  # Standardise text values.
df["city"] = df["city"].astype("string").str.strip()  # Standardise text values.
df["province"] = df["province"].astype("string").str.strip()  # Standardise text values.
df["region"] = df["region"].astype("string").str.strip()  # Standardise text values.
df["store_format"] = df["store_format"].astype("string").str.strip()  # Standardise text values.
df["opening_date"] = df["opening_date"].astype("string").str.strip()  # Standardise text values.
df["store_status"] = df["store_status"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["store_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/stores_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/stores_quality_profile.csv", index=False)  # Save quality profile.
```
