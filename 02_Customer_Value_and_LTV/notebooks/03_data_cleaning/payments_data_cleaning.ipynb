```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/payments.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["payment_date"] = pd.to_datetime(df["payment_date"], errors="coerce")  # Convert date values safely.
df["payment_status"] = pd.to_datetime(df["payment_status"], errors="coerce")  # Convert date values safely.
df["amount"] = pd.to_numeric(df["amount"], errors="coerce")  # Standardise numeric values.
df["refund_amount"] = pd.to_numeric(df["refund_amount"], errors="coerce")  # Standardise numeric values.
df["payment_id"] = df["payment_id"].astype("string").str.strip()  # Standardise text values.
df["order_id"] = df["order_id"].astype("string").str.strip()  # Standardise text values.
df["customer_id"] = df["customer_id"].astype("string").str.strip()  # Standardise text values.
df["payment_date"] = df["payment_date"].astype("string").str.strip()  # Standardise text values.
df["payment_method"] = df["payment_method"].astype("string").str.strip()  # Standardise text values.
df["payment_provider"] = df["payment_provider"].astype("string").str.strip()  # Standardise text values.
df["payment_status"] = df["payment_status"].astype("string").str.strip()  # Standardise text values.
df["transaction_reference"] = df["transaction_reference"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["payment_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/payments_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/payments_quality_profile.csv", index=False)  # Save quality profile.
```
