```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/suppliers_clean.csv")  # Load the cleaned dataset.
date_columns = ['contract_start_date', 'contract_status']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["payment_terms_days_missing_flag"] = df["payment_terms_days"].isna().astype(int)  # Flag missing numeric values.
df["supplier_id_present_flag"] = df["supplier_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/suppliers_features.csv", index=False)  # Save engineered dataset.
```
