```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/suppliers.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["contract_start_date"] = pd.to_datetime(df["contract_start_date"], errors="coerce")  # Convert date values safely.
df["contract_status"] = pd.to_datetime(df["contract_status"], errors="coerce")  # Convert date values safely.
df["payment_terms_days"] = pd.to_numeric(df["payment_terms_days"], errors="coerce")  # Standardise numeric values.
df["supplier_id"] = df["supplier_id"].astype("string").str.strip()  # Standardise text values.
df["supplier_name"] = df["supplier_name"].astype("string").str.strip()  # Standardise text values.
df["supplier_country"] = df["supplier_country"].astype("string").str.strip()  # Standardise text values.
df["supplier_type"] = df["supplier_type"].astype("string").str.strip()  # Standardise text values.
df["contract_start_date"] = df["contract_start_date"].astype("string").str.strip()  # Standardise text values.
df["contract_status"] = df["contract_status"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["supplier_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/suppliers_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/suppliers_quality_profile.csv", index=False)  # Save quality profile.
```
