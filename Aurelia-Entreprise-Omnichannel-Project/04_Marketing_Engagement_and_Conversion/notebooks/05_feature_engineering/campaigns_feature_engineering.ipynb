```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/campaigns_clean.csv")  # Load the cleaned dataset.
date_columns = ['start_date', 'end_date']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["budget_missing_flag"] = df["budget"].isna().astype(int)  # Flag missing numeric values.
df["campaign_id_present_flag"] = df["campaign_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/campaigns_features.csv", index=False)  # Save engineered dataset.
```
