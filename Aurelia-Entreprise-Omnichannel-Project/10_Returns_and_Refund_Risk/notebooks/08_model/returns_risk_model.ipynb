```python
import pandas as pd  # Import pandas for model preparation.
from sklearn.model_selection import train_test_split  # Import train-test splitting.
from sklearn.ensemble import RandomForestClassifier  # Import the classification model.
from sklearn.metrics import classification_report, accuracy_score  # Import evaluation metrics.
summary = pd.read_csv("outputs/business_eda_summary.csv")  # Load business EDA results.
summary["return_risk_flag"] = (summary.iloc[:, -1] >= summary.iloc[:, -1].median()).astype(int)  # Create return-risk target.
target = "return_risk_flag"  # Define the model target column.
features = summary.select_dtypes(include="number").drop(columns=[target], errors="ignore")  # Select numeric predictors.
features = features.fillna(0)  # Replace missing predictor values.
labels = summary[target]  # Extract target labels.
X_train, X_test, y_train, y_test = train_test_split(features, labels, test_size=0.2, random_state=42, stratify=labels if labels.nunique() > 1 else None)  # Split the dataset.
model = RandomForestClassifier(n_estimators=200, random_state=42)  # Initialise the random forest.
model.fit(X_train, y_train)  # Train the model.
predictions = model.predict(X_test)  # Generate test predictions.
metrics = pd.DataFrame({"metric":["accuracy"],"value":[accuracy_score(y_test, predictions)]})  # Calculate accuracy.
metrics.to_csv("outputs/model_metrics.csv", index=False)  # Save model metrics.
print(classification_report(y_test, predictions, zero_division=0))  # Print classification results.
```
