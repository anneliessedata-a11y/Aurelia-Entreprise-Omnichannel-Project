```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/products.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["category"] = pd.to_datetime(df["category"], errors="coerce")  # Convert date values safely.
df["launch_date"] = pd.to_datetime(df["launch_date"], errors="coerce")  # Convert date values safely.
df["discontinue_date"] = pd.to_datetime(df["discontinue_date"], errors="coerce")  # Convert date values safely.
df["product_status"] = pd.to_datetime(df["product_status"], errors="coerce")  # Convert date values safely.
df["standard_cost"] = pd.to_numeric(df["standard_cost"], errors="coerce")  # Standardise numeric values.
df["list_price"] = pd.to_numeric(df["list_price"], errors="coerce")  # Standardise numeric values.
df["product_id"] = df["product_id"].astype("string").str.strip()  # Standardise text values.
df["legacy_product_code"] = df["legacy_product_code"].astype("string").str.strip()  # Standardise text values.
df["sku"] = df["sku"].astype("string").str.strip()  # Standardise text values.
df["product_name"] = df["product_name"].astype("string").str.strip()  # Standardise text values.
df["brand"] = df["brand"].astype("string").str.strip()  # Standardise text values.
df["department"] = df["department"].astype("string").str.strip()  # Standardise text values.
df["category"] = df["category"].astype("string").str.strip()  # Standardise text values.
df["supplier_id"] = df["supplier_id"].astype("string").str.strip()  # Standardise text values.
df["launch_date"] = df["launch_date"].astype("string").str.strip()  # Standardise text values.
df["discontinue_date"] = df["discontinue_date"].astype("string").str.strip()  # Standardise text values.
df["product_status"] = df["product_status"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["product_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/products_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/products_quality_profile.csv", index=False)  # Save quality profile.
```
