```python
import pandas as pd  # Import pandas for business analysis.
import matplotlib.pyplot as plt  # Import Matplotlib for visualisation.
products_df = pd.read_csv("raw_data/products.csv")  # Load products business data.
order_items_df = pd.read_csv("raw_data/order_items.csv")  # Load order_items business data.
orders_df = pd.read_csv("raw_data/orders.csv")  # Load orders business data.
returns_df = pd.read_csv("raw_data/returns.csv")  # Load returns business data.
order_items_df["revenue"] = order_items_df["quantity"] * order_items_df["unit_price"]  # Calculate item revenue.
order_items_df["cost"] = order_items_df["quantity"] * order_items_df["unit_cost"]  # Calculate item cost.
order_items_df["margin"] = order_items_df["revenue"] - order_items_df["cost"]  # Calculate item margin.
summary = order_items_df.groupby("product_id")["margin"].agg(["count","mean","sum"]).reset_index()  # Summarise product margin.
summary["high_margin_flag"] = (summary["mean"] >= summary["mean"].median()).astype(int)  # Flag higher-margin products.
summary.to_csv("outputs/business_eda_summary.csv", index=False)  # Save business EDA summary.
print(summary.head(20))  # Display leading business findings.
```
