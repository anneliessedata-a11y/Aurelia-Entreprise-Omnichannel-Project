```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/inventory_daily_clean.csv")  # Load the cleaned dataset.
date_columns = ['inventory_date', 'location_id']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["opening_stock_missing_flag"] = df["opening_stock"].isna().astype(int)  # Flag missing numeric values.
df["stock_received_missing_flag"] = df["stock_received"].isna().astype(int)  # Flag missing numeric values.
df["units_sold_missing_flag"] = df["units_sold"].isna().astype(int)  # Flag missing numeric values.
df["units_reserved_missing_flag"] = df["units_reserved"].isna().astype(int)  # Flag missing numeric values.
df["product_id_present_flag"] = df["product_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/inventory_daily_features.csv", index=False)  # Save engineered dataset.
```
