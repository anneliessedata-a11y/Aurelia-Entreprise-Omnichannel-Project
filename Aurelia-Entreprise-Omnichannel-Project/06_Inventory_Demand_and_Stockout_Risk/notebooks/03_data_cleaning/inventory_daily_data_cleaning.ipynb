```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/inventory_daily.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["inventory_date"] = pd.to_datetime(df["inventory_date"], errors="coerce")  # Convert date values safely.
df["location_id"] = pd.to_datetime(df["location_id"], errors="coerce")  # Convert date values safely.
df["opening_stock"] = pd.to_numeric(df["opening_stock"], errors="coerce")  # Standardise numeric values.
df["stock_received"] = pd.to_numeric(df["stock_received"], errors="coerce")  # Standardise numeric values.
df["units_sold"] = pd.to_numeric(df["units_sold"], errors="coerce")  # Standardise numeric values.
df["units_reserved"] = pd.to_numeric(df["units_reserved"], errors="coerce")  # Standardise numeric values.
df["stock_adjustment"] = pd.to_numeric(df["stock_adjustment"], errors="coerce")  # Standardise numeric values.
df["closing_stock"] = pd.to_numeric(df["closing_stock"], errors="coerce")  # Standardise numeric values.
df["inventory_date"] = df["inventory_date"].astype("string").str.strip()  # Standardise text values.
df["product_id"] = df["product_id"].astype("string").str.strip()  # Standardise text values.
df["location_id"] = df["location_id"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["product_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/inventory_daily_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/inventory_daily_quality_profile.csv", index=False)  # Save quality profile.
```
