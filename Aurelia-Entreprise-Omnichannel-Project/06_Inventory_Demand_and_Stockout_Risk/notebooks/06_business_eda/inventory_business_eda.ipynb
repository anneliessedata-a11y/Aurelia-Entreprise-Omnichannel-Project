```python
import pandas as pd  # Import pandas for business analysis.
import matplotlib.pyplot as plt  # Import Matplotlib for visualisation.
inventory_daily_df = pd.read_csv("raw_data/inventory_daily.csv")  # Load inventory_daily business data.
products_df = pd.read_csv("raw_data/products.csv")  # Load products business data.
stores_df = pd.read_csv("raw_data/stores.csv")  # Load stores business data.
order_items_df = pd.read_csv("raw_data/order_items.csv")  # Load order_items business data.
orders_df = pd.read_csv("raw_data/orders.csv")  # Load orders business data.
inventory_daily_df["closing_stock"] = pd.to_numeric(inventory_daily_df["closing_stock"], errors="coerce")  # Standardise closing stock.
inventory_daily_df["reorder_point"] = pd.to_numeric(inventory_daily_df["reorder_point"], errors="coerce")  # Standardise reorder points.
inventory_daily_df["stockout_risk_flag"] = (inventory_daily_df["closing_stock"] <= inventory_daily_df["reorder_point"]).astype(int)  # Flag stockout risk.
summary = inventory_daily_df.groupby("product_id")["stockout_risk_flag"].mean().reset_index()  # Summarise product risk.
summary.to_csv("outputs/business_eda_summary.csv", index=False)  # Save business EDA summary.
print(summary.head(20))  # Display leading business findings.
```
