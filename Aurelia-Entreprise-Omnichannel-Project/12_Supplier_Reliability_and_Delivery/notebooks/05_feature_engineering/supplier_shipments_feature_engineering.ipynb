```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/supplier_shipments_clean.csv")  # Load the cleaned dataset.
date_columns = ['purchase_order_date', 'expected_delivery_date', 'actual_delivery_date', 'shipment_status']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["ordered_quantity_missing_flag"] = df["ordered_quantity"].isna().astype(int)  # Flag missing numeric values.
df["received_quantity_missing_flag"] = df["received_quantity"].isna().astype(int)  # Flag missing numeric values.
df["rejected_quantity_missing_flag"] = df["rejected_quantity"].isna().astype(int)  # Flag missing numeric values.
df["shipment_id_present_flag"] = df["shipment_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/supplier_shipments_features.csv", index=False)  # Save engineered dataset.
```
