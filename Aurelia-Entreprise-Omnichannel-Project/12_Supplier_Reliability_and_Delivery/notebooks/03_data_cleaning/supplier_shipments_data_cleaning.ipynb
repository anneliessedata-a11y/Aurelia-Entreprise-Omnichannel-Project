```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/supplier_shipments.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["purchase_order_date"] = pd.to_datetime(df["purchase_order_date"], errors="coerce")  # Convert date values safely.
df["expected_delivery_date"] = pd.to_datetime(df["expected_delivery_date"], errors="coerce")  # Convert date values safely.
df["actual_delivery_date"] = pd.to_datetime(df["actual_delivery_date"], errors="coerce")  # Convert date values safely.
df["shipment_status"] = pd.to_datetime(df["shipment_status"], errors="coerce")  # Convert date values safely.
df["ordered_quantity"] = pd.to_numeric(df["ordered_quantity"], errors="coerce")  # Standardise numeric values.
df["received_quantity"] = pd.to_numeric(df["received_quantity"], errors="coerce")  # Standardise numeric values.
df["rejected_quantity"] = pd.to_numeric(df["rejected_quantity"], errors="coerce")  # Standardise numeric values.
df["shipment_id"] = df["shipment_id"].astype("string").str.strip()  # Standardise text values.
df["supplier_id"] = df["supplier_id"].astype("string").str.strip()  # Standardise text values.
df["product_id"] = df["product_id"].astype("string").str.strip()  # Standardise text values.
df["purchase_order_date"] = df["purchase_order_date"].astype("string").str.strip()  # Standardise text values.
df["expected_delivery_date"] = df["expected_delivery_date"].astype("string").str.strip()  # Standardise text values.
df["actual_delivery_date"] = df["actual_delivery_date"].astype("string").str.strip()  # Standardise text values.
df["shipment_status"] = df["shipment_status"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["shipment_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/supplier_shipments_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/supplier_shipments_quality_profile.csv", index=False)  # Save quality profile.
```
