```python
import pandas as pd  # Import pandas for business analysis.
import matplotlib.pyplot as plt  # Import Matplotlib for visualisation.
suppliers_df = pd.read_csv("raw_data/suppliers.csv")  # Load suppliers business data.
supplier_shipments_df = pd.read_csv("raw_data/supplier_shipments.csv")  # Load supplier_shipments business data.
products_df = pd.read_csv("raw_data/products.csv")  # Load products business data.
supplier_shipments_df["expected_delivery_date"] = pd.to_datetime(supplier_shipments_df["expected_delivery_date"], errors="coerce")  # Convert expected dates.
supplier_shipments_df["actual_delivery_date"] = pd.to_datetime(supplier_shipments_df["actual_delivery_date"], errors="coerce")  # Convert actual dates.
supplier_shipments_df["late_days"] = (supplier_shipments_df["actual_delivery_date"] - supplier_shipments_df["expected_delivery_date"]).dt.days  # Calculate delivery delay.
supplier_shipments_df["late_delivery_flag"] = (supplier_shipments_df["late_days"] > 0).astype(int)  # Flag late deliveries.
summary = supplier_shipments_df.groupby("supplier_id")["late_delivery_flag"].mean().reset_index()  # Summarise supplier reliability.
summary.to_csv("outputs/business_eda_summary.csv", index=False)  # Save business EDA summary.
print(summary.head(20))  # Display leading business findings.
```
