```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/order_items_clean.csv")  # Load the cleaned dataset.
date_columns = ['fulfillment_status']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["quantity_missing_flag"] = df["quantity"].isna().astype(int)  # Flag missing numeric values.
df["unit_price_missing_flag"] = df["unit_price"].isna().astype(int)  # Flag missing numeric values.
df["discount_amount_missing_flag"] = df["discount_amount"].isna().astype(int)  # Flag missing numeric values.
df["unit_cost_missing_flag"] = df["unit_cost"].isna().astype(int)  # Flag missing numeric values.
df["order_item_id_present_flag"] = df["order_item_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/order_items_features.csv", index=False)  # Save engineered dataset.
```
