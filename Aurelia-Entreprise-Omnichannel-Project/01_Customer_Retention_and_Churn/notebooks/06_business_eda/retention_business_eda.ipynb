```python
import pandas as pd  # Import pandas for business analysis.
import matplotlib.pyplot as plt  # Import Matplotlib for visualisation.
customers_df = pd.read_csv("raw_data/customers.csv")  # Load customers business data.
orders_df = pd.read_csv("raw_data/orders.csv")  # Load orders business data.
payments_df = pd.read_csv("raw_data/payments.csv")  # Load payments business data.
returns_df = pd.read_csv("raw_data/returns.csv")  # Load returns business data.
support_tickets_df = pd.read_csv("raw_data/support_tickets.csv")  # Load support_tickets business data.
marketing_events_df = pd.read_csv("raw_data/marketing_events.csv")  # Load marketing_events business data.
orders_df["order_date"] = pd.to_datetime(orders_df["order_date"])  # Convert order dates.
last_order = orders_df.groupby("customer_id")["order_date"].max()  # Find each customer last order.
analysis = customers_df.merge(last_order.rename("last_order_date"), on="customer_id", how="left")  # Join activity to customers.
analysis["days_since_order"] = (analysis["last_order_date"].max() - analysis["last_order_date"]).dt.days  # Measure recency.
analysis["churn_flag"] = (analysis["days_since_order"] > analysis["days_since_order"].median()).astype(int)  # Create churn target.
summary = analysis.groupby("churn_flag")["customer_id"].count().reset_index(name="customers")  # Summarise retention groups.
summary.to_csv("outputs/business_eda_summary.csv", index=False)  # Save business EDA summary.
print(summary.head(20))  # Display leading business findings.
```
