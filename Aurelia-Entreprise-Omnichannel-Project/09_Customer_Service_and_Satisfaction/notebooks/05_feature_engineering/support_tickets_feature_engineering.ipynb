```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/support_tickets_clean.csv")  # Load the cleaned dataset.
date_columns = ['opened_at', 'closed_at', 'resolution_status', 'satisfaction_score']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["satisfaction_score_missing_flag"] = df["satisfaction_score"].isna().astype(int)  # Flag missing numeric values.
df["ticket_id_present_flag"] = df["ticket_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/support_tickets_features.csv", index=False)  # Save engineered dataset.
```
