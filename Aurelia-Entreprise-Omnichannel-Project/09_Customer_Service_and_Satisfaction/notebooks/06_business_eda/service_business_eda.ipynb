```python
import pandas as pd  # Import pandas for business analysis.
import matplotlib.pyplot as plt  # Import Matplotlib for visualisation.
support_tickets_df = pd.read_csv("raw_data/support_tickets.csv")  # Load support_tickets business data.
customers_df = pd.read_csv("raw_data/customers.csv")  # Load customers business data.
orders_df = pd.read_csv("raw_data/orders.csv")  # Load orders business data.
returns_df = pd.read_csv("raw_data/returns.csv")  # Load returns business data.
support_tickets_df["satisfaction_score"] = pd.to_numeric(support_tickets_df["satisfaction_score"], errors="coerce")  # Standardise satisfaction scores.
summary = support_tickets_df.groupby("issue_type")["satisfaction_score"].agg(["count","mean","median"]).reset_index()  # Summarise service satisfaction.
summary["low_satisfaction_flag"] = (summary["mean"] < summary["mean"].median()).astype(int)  # Flag lower satisfaction issues.
summary.to_csv("outputs/business_eda_summary.csv", index=False)  # Save business EDA summary.
print(summary.head(20))  # Display leading business findings.
```
