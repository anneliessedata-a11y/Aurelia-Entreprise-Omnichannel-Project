```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/stores_clean.csv")  # Load the cleaned dataset.
date_columns = ['store_format', 'opening_date', 'closing_date', 'store_status']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["closing_date_missing_flag"] = df["closing_date"].isna().astype(int)  # Flag missing numeric values.
df["floor_area_sqm_missing_flag"] = df["floor_area_sqm"].isna().astype(int)  # Flag missing numeric values.
df["store_id_present_flag"] = df["store_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/stores_features.csv", index=False)  # Save engineered dataset.
```
