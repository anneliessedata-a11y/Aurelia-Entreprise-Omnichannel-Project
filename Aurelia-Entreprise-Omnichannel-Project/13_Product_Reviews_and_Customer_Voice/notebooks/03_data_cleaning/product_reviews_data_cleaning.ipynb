```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/product_reviews.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["review_date"] = pd.to_datetime(df["review_date"], errors="coerce")  # Convert date values safely.
df["rating"] = pd.to_datetime(df["rating"], errors="coerce")  # Convert date values safely.
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")  # Standardise numeric values.
df["review_id"] = df["review_id"].astype("string").str.strip()  # Standardise text values.
df["customer_id"] = df["customer_id"].astype("string").str.strip()  # Standardise text values.
df["product_id"] = df["product_id"].astype("string").str.strip()  # Standardise text values.
df["order_item_id"] = df["order_item_id"].astype("string").str.strip()  # Standardise text values.
df["review_date"] = df["review_date"].astype("string").str.strip()  # Standardise text values.
df["review_title"] = df["review_title"].astype("string").str.strip()  # Standardise text values.
df["review_text"] = df["review_text"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["review_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/product_reviews_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/product_reviews_quality_profile.csv", index=False)  # Save quality profile.
```
