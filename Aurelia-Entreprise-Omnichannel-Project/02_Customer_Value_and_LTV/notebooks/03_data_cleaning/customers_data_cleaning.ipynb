```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/customers.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["date_of_birth"] = pd.to_datetime(df["date_of_birth"], errors="coerce")  # Convert date values safely.
df["registration_date"] = pd.to_datetime(df["registration_date"], errors="coerce")  # Convert date values safely.
df["loyalty_status"] = pd.to_datetime(df["loyalty_status"], errors="coerce")  # Convert date values safely.
df["customer_status"] = pd.to_datetime(df["customer_status"], errors="coerce")  # Convert date values safely.
df["legacy_customer_id"] = pd.to_numeric(df["legacy_customer_id"], errors="coerce")  # Standardise numeric values.
df["phone"] = pd.to_numeric(df["phone"], errors="coerce")  # Standardise numeric values.
df["customer_id"] = df["customer_id"].astype("string").str.strip()  # Standardise text values.
df["first_name"] = df["first_name"].astype("string").str.strip()  # Standardise text values.
df["last_name"] = df["last_name"].astype("string").str.strip()  # Standardise text values.
df["date_of_birth"] = df["date_of_birth"].astype("string").str.strip()  # Standardise text values.
df["gender"] = df["gender"].astype("string").str.strip()  # Standardise text values.
df["email"] = df["email"].astype("string").str.strip()  # Standardise text values.
df["registration_date"] = df["registration_date"].astype("string").str.strip()  # Standardise text values.
df["acquisition_channel"] = df["acquisition_channel"].astype("string").str.strip()  # Standardise text values.
df["home_city"] = df["home_city"].astype("string").str.strip()  # Standardise text values.
df["province"] = df["province"].astype("string").str.strip()  # Standardise text values.
df["customer_segment"] = df["customer_segment"].astype("string").str.strip()  # Standardise text values.
df["loyalty_status"] = df["loyalty_status"].astype("string").str.strip()  # Standardise text values.
df["customer_status"] = df["customer_status"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["customer_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/customers_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/customers_quality_profile.csv", index=False)  # Save quality profile.
```
