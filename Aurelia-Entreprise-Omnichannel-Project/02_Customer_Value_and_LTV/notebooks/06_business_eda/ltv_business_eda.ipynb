```python
import pandas as pd  # Import pandas for business analysis.
import matplotlib.pyplot as plt  # Import Matplotlib for visualisation.
customers_df = pd.read_csv("raw_data/customers.csv")  # Load customers business data.
orders_df = pd.read_csv("raw_data/orders.csv")  # Load orders business data.
order_items_df = pd.read_csv("raw_data/order_items.csv")  # Load order_items business data.
payments_df = pd.read_csv("raw_data/payments.csv")  # Load payments business data.
returns_df = pd.read_csv("raw_data/returns.csv")  # Load returns business data.
orders_df["gross_order_value"] = pd.to_numeric(orders_df["gross_order_value"], errors="coerce")  # Standardise order value.
value = orders_df.groupby("customer_id")["gross_order_value"].sum().reset_index(name="customer_value")  # Aggregate customer value.
analysis = customers_df.merge(value, on="customer_id", how="left")  # Join value to customers.
analysis["customer_value"] = analysis["customer_value"].fillna(0)  # Fill customers without orders.
analysis["high_value_flag"] = (analysis["customer_value"] >= analysis["customer_value"].median()).astype(int)  # Create value target.
summary = analysis.groupby("high_value_flag")["customer_value"].agg(["count","mean","sum"]).reset_index()  # Summarise customer value.
summary.to_csv("outputs/business_eda_summary.csv", index=False)  # Save business EDA summary.
print(summary.head(20))  # Display leading business findings.
```
