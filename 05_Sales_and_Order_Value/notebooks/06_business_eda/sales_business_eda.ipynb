```python
import pandas as pd  # Import pandas for business analysis.
import matplotlib.pyplot as plt  # Import Matplotlib for visualisation.
orders_df = pd.read_csv("raw_data/orders.csv")  # Load orders business data.
order_items_df = pd.read_csv("raw_data/order_items.csv")  # Load order_items business data.
customers_df = pd.read_csv("raw_data/customers.csv")  # Load customers business data.
products_df = pd.read_csv("raw_data/products.csv")  # Load products business data.
stores_df = pd.read_csv("raw_data/stores.csv")  # Load stores business data.
orders_df["gross_order_value"] = pd.to_numeric(orders_df["gross_order_value"], errors="coerce")  # Standardise order values.
summary = orders_df.groupby("channel")["gross_order_value"].agg(["count","sum","mean"]).reset_index()  # Summarise sales by channel.
summary["high_order_value_flag"] = (summary["mean"] >= summary["mean"].median()).astype(int)  # Flag higher-value channels.
summary.to_csv("outputs/business_eda_summary.csv", index=False)  # Save business EDA summary.
print(summary.head(20))  # Display leading business findings.
```
