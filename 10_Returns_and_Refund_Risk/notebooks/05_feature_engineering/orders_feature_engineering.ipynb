```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/orders_clean.csv")  # Load the cleaned dataset.
date_columns = ['order_date', 'order_status']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["gross_order_value_missing_flag"] = df["gross_order_value"].isna().astype(int)  # Flag missing numeric values.
df["discount_amount_missing_flag"] = df["discount_amount"].isna().astype(int)  # Flag missing numeric values.
df["shipping_fee_missing_flag"] = df["shipping_fee"].isna().astype(int)  # Flag missing numeric values.
df["tax_amount_missing_flag"] = df["tax_amount"].isna().astype(int)  # Flag missing numeric values.
df["order_id_present_flag"] = df["order_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/orders_features.csv", index=False)  # Save engineered dataset.
```
