```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/campaigns.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["start_date"] = pd.to_datetime(df["start_date"], errors="coerce")  # Convert date values safely.
df["end_date"] = pd.to_datetime(df["end_date"], errors="coerce")  # Convert date values safely.
df["budget"] = pd.to_numeric(df["budget"], errors="coerce")  # Standardise numeric values.
df["campaign_id"] = df["campaign_id"].astype("string").str.strip()  # Standardise text values.
df["campaign_name"] = df["campaign_name"].astype("string").str.strip()  # Standardise text values.
df["campaign_type"] = df["campaign_type"].astype("string").str.strip()  # Standardise text values.
df["channel"] = df["channel"].astype("string").str.strip()  # Standardise text values.
df["start_date"] = df["start_date"].astype("string").str.strip()  # Standardise text values.
df["end_date"] = df["end_date"].astype("string").str.strip()  # Standardise text values.
df["target_segment"] = df["target_segment"].astype("string").str.strip()  # Standardise text values.
df["objective"] = df["objective"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["campaign_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/campaigns_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/campaigns_quality_profile.csv", index=False)  # Save quality profile.
```
