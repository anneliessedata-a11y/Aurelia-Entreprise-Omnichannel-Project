```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/product_reviews_clean.csv")  # Load the cleaned dataset.
date_columns = ['review_date', 'rating']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["rating_missing_flag"] = df["rating"].isna().astype(int)  # Flag missing numeric values.
df["review_id_present_flag"] = df["review_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/product_reviews_features.csv", index=False)  # Save engineered dataset.
```
