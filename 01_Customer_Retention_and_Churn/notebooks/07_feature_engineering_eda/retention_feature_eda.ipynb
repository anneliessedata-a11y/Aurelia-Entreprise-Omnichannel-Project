```python
import pandas as pd  # Import pandas for feature analysis.
import matplotlib.pyplot as plt  # Import Matplotlib for visualisation.
summary = pd.read_csv("outputs/business_eda_summary.csv")  # Load business EDA output.
numeric = summary.select_dtypes(include="number")  # Select numerical features.
correlations = numeric.corr()  # Calculate feature correlations.
correlations.to_csv("outputs/feature_eda_correlations.csv")  # Save feature correlations.
print(correlations)  # Display feature relationships.
```
