```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/marketing_events.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["event_timestamp"] = pd.to_datetime(df["event_timestamp"], errors="coerce")  # Convert date values safely.
df["event_id"] = df["event_id"].astype("string").str.strip()  # Standardise text values.
df["customer_id"] = df["customer_id"].astype("string").str.strip()  # Standardise text values.
df["campaign_id"] = df["campaign_id"].astype("string").str.strip()  # Standardise text values.
df["event_timestamp"] = df["event_timestamp"].astype("string").str.strip()  # Standardise text values.
df["event_type"] = df["event_type"].astype("string").str.strip()  # Standardise text values.
df["device_type"] = df["device_type"].astype("string").str.strip()  # Standardise text values.
df["channel"] = df["channel"].astype("string").str.strip()  # Standardise text values.
df["session_id"] = df["session_id"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["event_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/marketing_events_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/marketing_events_quality_profile.csv", index=False)  # Save quality profile.
```
