```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/support_tickets.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["opened_at"] = pd.to_datetime(df["opened_at"], errors="coerce")  # Convert date values safely.
df["closed_at"] = pd.to_datetime(df["closed_at"], errors="coerce")  # Convert date values safely.
df["resolution_status"] = pd.to_datetime(df["resolution_status"], errors="coerce")  # Convert date values safely.
df["satisfaction_score"] = pd.to_datetime(df["satisfaction_score"], errors="coerce")  # Convert date values safely.
df["satisfaction_score"] = pd.to_numeric(df["satisfaction_score"], errors="coerce")  # Standardise numeric values.
df["ticket_id"] = df["ticket_id"].astype("string").str.strip()  # Standardise text values.
df["customer_id"] = df["customer_id"].astype("string").str.strip()  # Standardise text values.
df["opened_at"] = df["opened_at"].astype("string").str.strip()  # Standardise text values.
df["closed_at"] = df["closed_at"].astype("string").str.strip()  # Standardise text values.
df["issue_type"] = df["issue_type"].astype("string").str.strip()  # Standardise text values.
df["channel"] = df["channel"].astype("string").str.strip()  # Standardise text values.
df["priority"] = df["priority"].astype("string").str.strip()  # Standardise text values.
df["agent_id"] = df["agent_id"].astype("string").str.strip()  # Standardise text values.
df["resolution_status"] = df["resolution_status"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["ticket_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/support_tickets_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/support_tickets_quality_profile.csv", index=False)  # Save quality profile.
```
