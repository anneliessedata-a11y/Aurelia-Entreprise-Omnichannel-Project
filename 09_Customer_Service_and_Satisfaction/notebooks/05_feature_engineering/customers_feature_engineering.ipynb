```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/customers_clean.csv")  # Load the cleaned dataset.
date_columns = ['date_of_birth', 'registration_date', 'loyalty_status', 'customer_status']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["legacy_customer_id_missing_flag"] = df["legacy_customer_id"].isna().astype(int)  # Flag missing numeric values.
df["phone_missing_flag"] = df["phone"].isna().astype(int)  # Flag missing numeric values.
df["customer_id_present_flag"] = df["customer_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/customers_features.csv", index=False)  # Save engineered dataset.
```
