```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/products_clean.csv")  # Load the cleaned dataset.
date_columns = ['category', 'launch_date', 'discontinue_date', 'product_status']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["standard_cost_missing_flag"] = df["standard_cost"].isna().astype(int)  # Flag missing numeric values.
df["list_price_missing_flag"] = df["list_price"].isna().astype(int)  # Flag missing numeric values.
df["product_id_present_flag"] = df["product_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/products_features.csv", index=False)  # Save engineered dataset.
```
