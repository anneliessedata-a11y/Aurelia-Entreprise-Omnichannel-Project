```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/returns.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["return_request_date"] = pd.to_datetime(df["return_request_date"], errors="coerce")  # Convert date values safely.
df["return_received_date"] = pd.to_datetime(df["return_received_date"], errors="coerce")  # Convert date values safely.
df["refund_status"] = pd.to_datetime(df["refund_status"], errors="coerce")  # Convert date values safely.
df["refund_amount"] = pd.to_numeric(df["refund_amount"], errors="coerce")  # Standardise numeric values.
df["return_id"] = df["return_id"].astype("string").str.strip()  # Standardise text values.
df["order_item_id"] = df["order_item_id"].astype("string").str.strip()  # Standardise text values.
df["customer_id"] = df["customer_id"].astype("string").str.strip()  # Standardise text values.
df["return_request_date"] = df["return_request_date"].astype("string").str.strip()  # Standardise text values.
df["return_received_date"] = df["return_received_date"].astype("string").str.strip()  # Standardise text values.
df["return_reason"] = df["return_reason"].astype("string").str.strip()  # Standardise text values.
df["return_condition"] = df["return_condition"].astype("string").str.strip()  # Standardise text values.
df["refund_status"] = df["refund_status"].astype("string").str.strip()  # Standardise text values.
df["return_channel"] = df["return_channel"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["return_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/returns_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/returns_quality_profile.csv", index=False)  # Save quality profile.
```
