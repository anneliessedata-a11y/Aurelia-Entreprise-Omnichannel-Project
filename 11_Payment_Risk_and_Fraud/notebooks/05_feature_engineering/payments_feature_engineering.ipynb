```python
import pandas as pd  # Import pandas for feature engineering.
df = pd.read_csv("outputs/payments_clean.csv")  # Load the cleaned dataset.
date_columns = ['payment_date', 'payment_status']  # Identify date fields for features.
for column in date_columns:  # Process each date field.
    df[f"{column}_year"] = pd.to_datetime(df[column], errors="coerce").dt.year  # Extract calendar year.
    df[f"{column}_month"] = pd.to_datetime(df[column], errors="coerce").dt.month  # Extract calendar month.
df["amount_missing_flag"] = df["amount"].isna().astype(int)  # Flag missing numeric values.
df["refund_amount_missing_flag"] = df["refund_amount"].isna().astype(int)  # Flag missing numeric values.
df["payment_id_present_flag"] = df["payment_id"].notna().astype(int)  # Flag valid identifiers.
df.to_csv("outputs/payments_features.csv", index=False)  # Save engineered dataset.
```
