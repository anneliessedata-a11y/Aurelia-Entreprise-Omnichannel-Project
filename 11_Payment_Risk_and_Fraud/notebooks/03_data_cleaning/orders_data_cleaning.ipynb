```python
import pandas as pd  # Import pandas for data manipulation.
import numpy as np  # Import NumPy for numerical operations.
df = pd.read_csv("raw_data/orders.csv")  # Load the raw dataset.
df = df.drop_duplicates()  # Remove duplicate records.
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")  # Convert date values safely.
df["order_status"] = pd.to_datetime(df["order_status"], errors="coerce")  # Convert date values safely.
df["gross_order_value"] = pd.to_numeric(df["gross_order_value"], errors="coerce")  # Standardise numeric values.
df["discount_amount"] = pd.to_numeric(df["discount_amount"], errors="coerce")  # Standardise numeric values.
df["shipping_fee"] = pd.to_numeric(df["shipping_fee"], errors="coerce")  # Standardise numeric values.
df["tax_amount"] = pd.to_numeric(df["tax_amount"], errors="coerce")  # Standardise numeric values.
df["order_id"] = df["order_id"].astype("string").str.strip()  # Standardise text values.
df["customer_id"] = df["customer_id"].astype("string").str.strip()  # Standardise text values.
df["order_date"] = df["order_date"].astype("string").str.strip()  # Standardise text values.
df["channel"] = df["channel"].astype("string").str.strip()  # Standardise text values.
df["store_id"] = df["store_id"].astype("string").str.strip()  # Standardise text values.
df["fulfillment_method"] = df["fulfillment_method"].astype("string").str.strip()  # Standardise text values.
df["order_status"] = df["order_status"].astype("string").str.strip()  # Standardise text values.
df["payment_method"] = df["payment_method"].astype("string").str.strip()  # Standardise text values.
df = df.dropna(subset=["order_id"])  # Remove rows missing the primary key.
df.to_csv("outputs/orders_clean.csv", index=False)  # Save cleaned dataset.
profile = pd.DataFrame({"column": df.columns, "missing": df.isna().sum(), "unique": df.nunique()})  # Build quality profile.
profile.to_csv("outputs/orders_quality_profile.csv", index=False)  # Save quality profile.
```
